```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
from scipy.special import jn_zeros

# Evaluate the supplied integral for the Bessel function J_n(x).
def J(n, x):
    integral, error = quad(lambda theta: np.cos(x * np.sin(theta) - n * theta),
                           0, np.pi)
    return integral / np.pi

# Sample the domain, including both endpoints.
x_values = np.linspace(0, 20, 1000)

plt.figure(figsize=(10, 6))
for n, color in zip([0, 1, 2], ['tab:blue', 'tab:orange', 'tab:green']):
    y_values = [J(n, x) for x in x_values]
    plt.plot(x_values, y_values, color=color, label=f'n = {n}')

    # Mark the first five positive roots on the x-axis in the curve's color.
    roots = jn_zeros(n, 5)
    plt.scatter(roots, np.zeros_like(roots), color=color, s=50,
                edgecolors='white', linewidths=0.7, zorder=5)


# Include all three curves with some room above and below their extrema.
plt.xlim(0, 20)
plt.ylim(-0.5, 1.1)
plt.xlabel('x')
plt.ylabel(r'$J_n(x)$')
plt.title('Bessel Functions for n = 0, 1, 2')
plt.axhline(0, color='gray', linewidth=0.7)
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from scipy.special import jn_zeros
import pandas as pd

# Find the first five positive zeros of J_n(x) for each order n.
# The root at x = 0 for n = 1 and n = 2 is excluded.
root_rows = []
for n in [0, 1, 2]:
    roots = jn_zeros(n, 5)
    root_rows.append([n, *roots])

# Show one row per n, with the roots in increasing order.
root_columns = [f'Root {i}' for i in range(1, 6)]
roots_table = pd.DataFrame(root_rows, columns=['n', *root_columns])
roots_table.style.format({column: '{:.6f}' for column in root_columns}).hide(axis='index')
```
